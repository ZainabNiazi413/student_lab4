```python
import time, timeit, tracemalloc

N = 1_000_000
py_list = list(range(N))
np_arr = np.arange(N)

def loop_double(values):
    result = []
    for x in values:
        result.append(x * 2)
    return result

def numpy_double(values):
    return values * 2
```

```python
for label, func, data in (('loop', loop_double, py_list), ('numpy', numpy_double, np_arr)):
    start = time.perf_counter()
    func(data)
    print(label, round(time.perf_counter() - start, 5), 'seconds')
```

```python
n = 20
t_loop = timeit.timeit('loop_double(py_list)', globals=globals(), number=n) / n
t_np = timeit.timeit('numpy_double(np_arr)', globals=globals(), number=n) / n

print('loop', round(t_loop, 5), 'numpy', round(t_np, 5), 'ratio', round(t_loop / t_np))
```

```python
for label, make in (('list', lambda: [x * 2 for x in range(N)]),
                    ('array', lambda: np.arange(N) * 2)):
    tracemalloc.start()
    obj = make()
    current, peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    print(label, 'current MB', round(current / 1e6, 1), 'peak MB', round(peak / 1e6, 1))
```

```python
sizes = [1_000, 100_000, 1_000_000]
rows = []

for N_test in sizes:
    py_test = list(range(N_test))
    np_test = np.arange(N_test)

    n = 100 if N_test == 1_000 else 10

    t_loop = timeit.timeit(
        lambda: loop_double(py_test), number=n
    ) / n

    t_np = timeit.timeit(
        lambda: numpy_double(np_test), number=n
    ) / n

    ratio = t_loop / t_np
    rows.append((N_test, t_loop, t_np, ratio))

print(f"{'N':>10} {'Loop (s)':>12} {'NumPy (s)':>12} {'Ratio':>10}")

for N_test, t_loop, t_np, ratio in rows:
    print(f"{N_test:>10} {t_loop:>12.6f} {t_np:>12.6f} {ratio:>10.1f}")
```

```python
N = 1_000_000
py_list = list(range(N))
np_arr = np.arange(N)

n = 10

t_loop = min(timeit.repeat(
    lambda: loop_double(py_list), number=n, repeat=5
)) / n

t_np = min(timeit.repeat(
    lambda: numpy_double(np_arr), number=n, repeat=5
)) / n

print("Best loop time:", round(t_loop, 6), "seconds")
print("Best NumPy time:", round(t_np, 6), "seconds")
print("Ratio:", round(t_loop / t_np, 1))
```
