```python
import numpy as np

# part 1
ones_arr = np.ones((2, 3))
full_arr = np.full((2, 2), 7)
evens_arr = np.arange(2, 11, 2)  # 2 to 10 inclusive, step of 2

print("Ones array:", ones_arr)
print("Full array:", full_arr)
print("Evens array:", evens_arr)

# part 2
arr_3d = np.zeros((2, 3, 4))
print("ndim:", arr_3d.ndim)
print("shape:", arr_3d.shape)
print("size:", arr_3d.size)

# part 3
mixed_arr = np.array([1, 2.5, 3])
print("dtype:", mixed_arr.dtype)
```
